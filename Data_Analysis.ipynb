Fill the nulkl values with median.

```python
data['Review Rating'] = (
    data.groupby('Category')['Review Rating']
        .transform(lambda x: x.fillna(x.median()))
)
```

```python
data.isnull().sum()
```

Data Pre-processing

```python
data.columns = (
    data.columns
        .str.lower()
        .str.replace(' ', '_')
)
```

```python
data.isnull().sum()
```

Data Cleaning

```python
data.columns = (
    data.columns
        .str.lower()
        .str.replace(r'[^\w]', '_', regex=True)
        .str.replace(r'_+', '_', regex=True)
        .str.strip('_')
)
```

```python
# @title
data.columns
```

Sampling Generations

```python
labels = ['Young', 'Adult', 'Middle-Aged', 'Senior']

data['age_group'] = pd.qcut(
    data['age'],
    q=4,
    labels=labels
)
```

```python
data.columns
```

```python
frequency_mapping={
    'Fortinightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90,
}
data['frequency_of_purchases']=data['frequency_of_purchases'].map(frequency_mapping)
```
